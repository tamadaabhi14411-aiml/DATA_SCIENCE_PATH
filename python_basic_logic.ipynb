```python
try:
    num_input = int(input("enter the numbers: ")) # Store the integer input here
    x = num_input # Assign the integer to x
    result = x / 100
    print(f"The result is: {result}") # Print the result here
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Invalid input. Please enter a number.")
finally:
    print("Program finished")
```

### Question: Even/Odd Counter

Write a Python function called `count_even_odd` that takes a list of integers as input. This function should iterate through the list and determine how many numbers are even and how many are odd. Finally, it should print the total count of even numbers and odd numbers.

After defining the function, prompt the user to enter a comma-separated list of numbers (e.g., `1,2,3,4,5`). Convert this input string into a list of integers. Use a `try-except` block to handle potential `ValueError` if the user enters non-numeric input. If the input is valid, call your `count_even_odd` function with the created list. If an error occurs, print an appropriate error message.


```python
def count_even_odd(numbers):
    even_count = 0
    odd_count = 0

    for num in numbers:
        if num % 2 == 0:
            even_count +=1
        else:
            odd_count +=1

    print("Even count:", even_count)
    print("Odd count:", odd_count)

covert = input("Enter a comma-separated list of numbers: ")

try:
    numbers = [int(num.strip()) for num in covert.split(",")]      #[expression for item in sequence]
    count_even_odd(numbers)
except ValueError:
    print("Invalid input. Please enter a comma-separated list of integers.")



```

```python
def count_even_odd(numbers_list):
    """
    Counts the number of even and odd integers in a list.
    Args:
        numbers_list (list): A list of integers.
    """
    even_count = 0
    odd_count = 0
    for num in numbers_list:
        if num % 2 == 0:
            even_count += 1
        else:
            odd_count += 1
    print(f"Total even numbers: {even_count}")
    print(f"Total odd numbers: {odd_count}")

# Get input from the user
user_input = input("Enter a comma-separated list of numbers (e.g., 1,2,3,4,5): ")

try:
    # Split the input string by comma and convert to integers
    numbers = [int(num.strip()) for num in user_input.split(',')]

    # Call the function with the list of numbers
    count_even_odd(numbers)
except ValueError:
    print("Invalid input. Please enter only comma-separated numbers.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
